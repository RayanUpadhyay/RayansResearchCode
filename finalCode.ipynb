```python
import pandas as pd
import numpy as np
from autogluon.tabular import TabularPredictor
from imblearn.over_sampling import SMOTE
import pprint

#gettind dataset, seperating based on whitespaces
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/statlog/german/german.data"
df = pd.read_csv(url, delim_whitespace=True, header=None)

df.columns = [
    "status_checking", "duration", "credit_history", "purpose", "credit_amount",
    "savings", "employment_since", "installment_rate", "personal_status_sex",
    "other_debtors", "residence_since", "property", "age", "other_installment_plans",
    "housing", "existing_credits", "job", "num_dependents", "telephone",
    "foreign_worker", "target"
]

df["target"] = df["target"].map({1: 0, 2: 1})  
'''
this dataset returns 1 as good credit risk and 2 as bad credit risk so I am mapping 1 to 0 and 2 to 1 so that it's binary classification
autogluon uses 0 and 1, and it can be found here: https://auto.gluon.ai/0.4.0/api/autogluon.predictor.html in the 
class_labels_internal section
'''

X = df.drop("target", axis=1) #ensures X is everything but the label
y = df["target"] #y becomes the column to predict

X_sensitive = X.copy() #keeps a copy of X before it is subjected to one hot encoding

#One hot encoding stage:
X_encoded = pd.get_dummies(X, drop_first=True)

#SMOTE Stage
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_encoded, y)

#ensuring train_data now has both the features and the target
train_data = pd.concat(
    [pd.DataFrame(X_resampled, columns=X_encoded.columns),
     pd.Series(y_resampled, name="target")],
    axis=1
)

#training autogluon with XGBOOST, RF, GDM, NN_TORCH, CAT
predictor = TabularPredictor(
    label='target',
    eval_metric='roc_auc'
).fit(
    train_data,
    hyperparameters={'XGB' : {}, 'RF' : {}, 'GBM' : {}, 'NN_TORCH' : {}, 'CAT' : {}}
)

#prediction stage
X_test = train_data.drop(columns=['target'])
y_true = train_data['target']
y_pred = predictor.predict(X_test)

#functions for fairness metrics
def equalized_odds_difference(y_true, y_pred, protected_attr): 
    df_temp = pd.DataFrame({
        'y_true': y_true,
        'y_pred': y_pred,
        'protected_attr': protected_attr
    })
    
    def calculate_rates(group_data):
        tp = ((group_data['y_true'] == 1) & (group_data['y_pred'] == 1)).sum()
        tn = ((group_data['y_true'] == 0) & (group_data['y_pred'] == 0)).sum()
        fp = ((group_data['y_true'] == 0) & (group_data['y_pred'] == 1)).sum()
        fn = ((group_data['y_true'] == 1) & (group_data['y_pred'] == 0)).sum()
        
        tpr = tp / (tp + fn) if (tp + fn) > 0 else 0
        fpr = fp / (fp + tn) if (fp + tn) > 0 else 0
        return tpr, fpr
    
    groups = df_temp['protected_attr'].unique()
    rates = {group: calculate_rates(df_temp[df_temp['protected_attr'] == group]) for group in groups}
    
    tprs = [rate[0] for rate in rates.values()]
    fprs = [rate[1] for rate in rates.values()]
    
    return {
        'TPR_per_group': {group: rate[0] for group, rate in rates.items()},
        'FPR_per_group': {group: rate[1] for group, rate in rates.items()},
        'TPR_difference': np.max(tprs) - np.min(tprs),
        'FPR_difference': np.max(fprs) - np.min(fprs)
    }

def disparate_impact(y_pred, protected_attr):    
    df_temp = pd.DataFrame({'y_pred': y_pred, 'protected_attr': protected_attr})
    groups = df_temp['protected_attr'].unique()
    impacts = {}
    
    base_group = groups[0]
    p_base = (df_temp.loc[df_temp['protected_attr'] == base_group, 'y_pred'] == 1).mean()
    
    for group in groups:
        p_group = (df_temp.loc[df_temp['protected_attr'] == group, 'y_pred'] == 1).mean()
        impacts[group] = p_group / p_base if p_base > 0 else np.nan
    
    return impacts

def predictive_parity(y_true, y_pred, group):
    df_temp = pd.DataFrame({'y_true': y_true, 'y_pred': y_pred, 'group': group})
    groups = df_temp['group'].unique()
    pp_per_group = {}
    
    for g in groups:
        pred_pos = df_temp[(df_temp['group'] == g) & (df_temp['y_pred'] == 1)]
        if len(pred_pos) > 0:
            pp_per_group[g] = pred_pos['y_true'].mean()
        else:
            pp_per_group[g] = np.nan
    
    pp_diff = np.nanmax(list(pp_per_group.values())) - np.nanmin(list(pp_per_group.values()))
    return {'PP_per_group': pp_per_group, 'PP_difference': pp_diff}

#evaluating fairness for every sensitive attribute
categorical_features = [
    "status_checking", "credit_history", "purpose", "savings", "employment_since",
    "personal_status_sex", "other_debtors", "property", "other_installment_plans",
    "housing", "job", "telephone", "foreign_worker"
]

fairness_results = {}

for feature in categorical_features:
    protected_attr = X_sensitive[feature]  # original categorical values
    fairness_results[feature] = {
        "Equalized_Odds": equalized_odds_difference(y_true, y_pred, protected_attr),
        "Disparate_Impact": disparate_impact(y_pred, protected_attr),
        "Predictive_Parity": predictive_parity(y_true, y_pred, protected_attr)
    }

#displaying of the results using PrettyPrinter(to make it easier to read)
pp = pprint.PrettyPrinter(indent=2)
pp.pprint(fairness_results)

import matplotlib.pyplot as plt
import seaborn as sns

#convert fairness_results into a format suitable for plotting
plot_data = []
for feature, metrics in fairness_results.items():
    #Equalized Odds
    plot_data.append({
        'Feature': feature,
        'Metric': 'TPR_difference',
        'Value': metrics['Equalized_Odds']['TPR_difference']
    })
    plot_data.append({
        'Feature': feature,
        'Metric': 'FPR_difference',
        'Value': metrics['Equalized_Odds']['FPR_difference']
    })
    #Disparate Impact (use ratio difference between groups)
    disparate_values = list(metrics['Disparate_Impact'].values())
    if len(disparate_values) > 1:
        di_diff = max(disparate_values) - min(disparate_values)
    else:
        di_diff = 0
    plot_data.append({
        'Feature': feature,
        'Metric': 'Disparate_Impact_Difference',
        'Value': di_diff
    })
    #Predictive Parity
    plot_data.append({
        'Feature': feature,
        'Metric': 'PP_difference',
        'Value': metrics['Predictive_Parity']['PP_difference']
    })

plot_df = pd.DataFrame(plot_data)

#plot each metric separately for clarity
metrics_to_plot = ['TPR_difference', 'FPR_difference', 'Disparate_Impact_Difference', 'PP_difference']

for metric in metrics_to_plot:
    plt.figure(figsize=(12,6))
    sns.barplot(data=plot_df[plot_df['Metric'] == metric],
                x='Feature', y='Value', palette='viridis')
    plt.title(f'Fairness Metric: {metric} across Features')
    plt.xticks(rotation=45, ha='right')
    plt.ylabel(metric)
    plt.tight_layout()
    plt.show()
```

```python
# Repeated-eval block: computes mean & std across repeated stratified splits
# (based on your existing code / variable names and fairness functions).
# Requires: scikit-learn, imblearn, autogluon (you already used), optional: ctgan/sdv for CTGAN/TVAE/Copula.

import warnings
warnings.filterwarnings('ignore')

import pandas as pd
import numpy as np
from sklearn.model_selection import RepeatedStratifiedKFold
from sklearn.metrics import roc_auc_score
import pprint
import time

# imports you already used
from imblearn.over_sampling import SMOTE
from autogluon.tabular import TabularPredictor

# -------------------------------------------------------
# --- Reuse / (slightly hardened) fairness metric funcs ---
# -------------------------------------------------------
def equalized_odds_difference(y_true, y_pred, protected_attr):
    df_temp = pd.DataFrame({'y_true': y_true, 'y_pred': y_pred, 'protected_attr': protected_attr})
    def calculate_rates(group_data):
        tp = ((group_data['y_true'] == 1) & (group_data['y_pred'] == 1)).sum()
        tn = ((group_data['y_true'] == 0) & (group_data['y_pred'] == 0)).sum()
        fp = ((group_data['y_true'] == 0) & (group_data['y_pred'] == 1)).sum()
        fn = ((group_data['y_true'] == 1) & (group_data['y_pred'] == 0)).sum()
        tpr = tp / (tp + fn) if (tp + fn) > 0 else np.nan
        fpr = fp / (fp + tn) if (fp + tn) > 0 else np.nan
        return tpr, fpr
    groups = df_temp['protected_attr'].unique()
    rates = {}
    for group in groups:
        grp = df_temp[df_temp['protected_attr'] == group]
        rates[group] = calculate_rates(grp)
    tprs = [r[0] for r in rates.values() if not np.isnan(r[0])]
    fprs = [r[1] for r in rates.values() if not np.isnan(r[1])]
    # if any group had nan (no positives/negatives) then differences become nan; handle gracefully
    tpr_diff = (np.nanmax(tprs) - np.nanmin(tprs)) if len(tprs) > 0 else np.nan
    fpr_diff = (np.nanmax(fprs) - np.nanmin(fprs)) if len(fprs) > 0 else np.nan
    return {'TPR_per_group': {g: rates[g][0] for g in rates}, 
            'FPR_per_group': {g: rates[g][1] for g in rates},
            'TPR_difference': tpr_diff,
            'FPR_difference': fpr_diff}

def disparate_impact(y_pred, protected_attr):
    df_temp = pd.DataFrame({'y_pred': y_pred, 'protected_attr': protected_attr})
    groups = df_temp['protected_attr'].unique()
    impacts = {}
    # choose a base group deterministically (smallest label string)
    groups_sorted = sorted(groups)
    base_group = groups_sorted[0]
    p_base = (df_temp.loc[df_temp['protected_attr'] == base_group, 'y_pred'] == 1).mean()
    for group in groups:
        p_group = (df_temp.loc[df_temp['protected_attr'] == group, 'y_pred'] == 1).mean()
        impacts[group] = (p_group / p_base) if p_base > 0 else np.nan
    # summarize as max distance from 1 (perfect parity)
    deviations = [abs(v - 1) for v in impacts.values() if not np.isnan(v)]
    max_dev = max(deviations) if deviations else np.nan
    return {'impacts_by_group': impacts, 'max_abs_deviation_from_1': max_dev}

def predictive_parity(y_true, y_pred, group):
    df_temp = pd.DataFrame({'y_true': y_true, 'y_pred': y_pred, 'group': group})
    groups = df_temp['group'].unique()
    pp_per_group = {}
    for g in groups:
        pred_pos = df_temp[(df_temp['group'] == g) & (df_temp['y_pred'] == 1)]
        if len(pred_pos) > 0:
            pp_per_group[g] = pred_pos['y_true'].mean()
        else:
            pp_per_group[g] = np.nan
    vals = [v for v in pp_per_group.values() if not np.isnan(v)]
    pp_diff = (max(vals) - min(vals)) if len(vals) > 0 else np.nan
    return {'PP_per_group': pp_per_group, 'PP_difference': pp_diff}

# -------------------------
# --- experiment config ---
# -------------------------
N_SPLITS = 5          # folds per repeat
N_REPEATS = 5         # repeats -> total runs = N_SPLITS * N_REPEATS
RANDOM_STATE = 42

# the protected (sensitive) columns you evaluate fairness over:
categorical_features = [
    "status_checking", "credit_history", "purpose", "savings", "employment_since",
    "personal_status_sex", "other_debtors", "property", "other_installment_plans",
    "housing", "job", "telephone", "foreign_worker"
]

# --- which synthetic methods to run (we always run 'baseline' and 'smote' if available) ---
methods_to_try = ['baseline', 'smote', 'ctgan', 'tvae', 'gaussian_copula']

# ---------------------------------------------------
# --- try importing optional synthesizer libraries ---
# ---------------------------------------------------
ctgan_available = False
tvae_available = False
copula_available = False
try:
    from ctgan import CTGAN  # older ctgan package
    ctgan_available = True
except Exception:
    try:
        # SDV provides CTGANSynthesizer/TVAESynthesizer in sdv.tabular (if installed)
        from sdv.tabular import CTGANSynthesizer, TVAESynthesizer
        ctgan_available = True
        tvae_available = True
    except Exception:
        ctgan_available = False
        tvae_available = False

try:
    # gaussian copula model often from sdv.rel or sdv.tabular.GaussianCopulaSynthesizer
    from sdv.tabular import GaussianCopulaSynthesizer
    copula_available = True
except Exception:
    copula_available = False

# remove methods that aren't available (but do not error)
if not ctgan_available and 'ctgan' in methods_to_try:
    methods_to_try.remove('ctgan')
if not tvae_available and 'tvae' in methods_to_try:
    methods_to_try.remove('tvae')
if not copula_available and 'gaussian_copula' in methods_to_try:
    methods_to_try.remove('gaussian_copula')

print("Methods that will be tested:", methods_to_try)

# -------------------------------------------------------
# --- Load dataset exactly as you do in the notebook ---
# -------------------------------------------------------
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/statlog/german/german.data"
df = pd.read_csv(url, delim_whitespace=True, header=None)

df.columns = [
    "status_checking", "duration", "credit_history", "purpose", "credit_amount",
    "savings", "employment_since", "installment_rate", "personal_status_sex",
    "other_debtors", "residence_since", "property", "age", "other_installment_plans",
    "housing", "existing_credits", "job", "num_dependents", "telephone",
    "foreign_worker", "target"
]
df["target"] = df["target"].map({1: 0, 2: 1})

X = df.drop("target", axis=1)
y = df["target"]
X_sensitive = X.copy()
X_encoded = pd.get_dummies(X, drop_first=True)

# get label prevalence (for Methodology section)
label_prevalence = y.value_counts(normalize=True).sort_index().to_dict()  # {0: p0, 1: p1}
print("Label prevalence (0=good,1=bad):", label_prevalence)

# -------------------------------------------------------
# --- repeated stratified splits and per-method runs ---
# -------------------------------------------------------
rskf = RepeatedStratifiedKFold(n_splits=N_SPLITS, n_repeats=N_REPEATS, random_state=RANDOM_STATE)

# storage: method -> list of per-run metric dicts
results_by_method = {m: [] for m in methods_to_try}

run_counter = 0
total_runs = N_SPLITS * N_REPEATS * len(methods_to_try)
start_time = time.time()

for fold_idx, (train_idx, test_idx) in enumerate(rskf.split(X_encoded, y)):
    run_counter += 1
    print(f"\n---- split {fold_idx+1}/{N_SPLITS * N_REPEATS}  (run group {run_counter}) ----")
    X_train_enc = X_encoded.iloc[train_idx].reset_index(drop=True)
    X_test_enc  = X_encoded.iloc[test_idx].reset_index(drop=True)
    y_train = y.iloc[train_idx].reset_index(drop=True)
    y_test  = y.iloc[test_idx].reset_index(drop=True)
    # keep protected groups for test (original categorical values)
    X_sensitive_test = X_sensitive.iloc[test_idx].reset_index(drop=True)
    
    for method in methods_to_try:
        print(" Running method:", method)
        # create training data according to method
        if method == 'baseline':
            X_tr = X_train_enc.copy()
            y_tr = y_train.copy()
        elif method == 'smote':
            sm = SMOTE(random_state=RANDOM_STATE)
            X_res, y_res = sm.fit_resample(X_train_enc, y_train)
            X_tr = pd.DataFrame(X_res, columns=X_train_enc.columns)
            y_tr = pd.Series(y_res, name='target')
        elif method == 'ctgan':
            # try to fit CTGAN on original (non-onehot ideally) but we have onehot; quick approach: fit on X_train_enc + target
            try:
                if ctgan_available:
                    # try sdv or ctgan import variants
                    try:
                        # sdv:
                        synth = CTGANSynthesizer()  # only if sdv.tabular import succeeded earlier
                        X_ctgan_train = X_train_enc.copy()
                        X_ctgan_train['target'] = y_train.values
                        synth.fit(X_ctgan_train)
                        X_synth = synth.sample(len(X_train_enc))
                    except Exception:
                        # fallback to ctgan.CTGAN
                        try:
                            from ctgan import CTGAN as _CTGAN
                            synth = _CTGAN(epochs=10)
                            X_ctgan_train = X_train_enc.copy()
                            X_ctgan_train['target'] = y_train.values
                            synth.fit(X_ctgan_train, discrete_columns=[c for c in X_ctgan_train.columns if X_ctgan_train[c].dtype == 'uint8'])
                            X_synth = synth.sample(len(X_train_enc))
                        except Exception as e:
                            print(" ctgan sampling failed:", e)
                            continue
                    # split synthetic
                    y_tr = X_synth['target']
                    X_tr = X_synth.drop(columns=['target'])
                else:
                    print(" ctgan not available; skipping")
                    continue
            except Exception as e:
                print(" ctgan error:", e)
                continue
        elif method == 'tvae':
            try:
                if tvae_available:
                    synth = TVAESynthesizer()
                    X_tvae_train = X_train_enc.copy()
                    X_tvae_train['target'] = y_train.values
                    synth.fit(X_tvae_train)
                    X_synth = synth.sample(len(X_train_enc))
                    y_tr = X_synth['target']
                    X_tr = X_synth.drop(columns=['target'])
                else:
                    print(" tvae not available; skipping")
                    continue
            except Exception as e:
                print(" tvae error:", e)
                continue
        elif method == 'gaussian_copula':
            try:
                if copula_available:
                    synth = GaussianCopulaSynthesizer()
                    X_gc_train = X_train_enc.copy()
                    X_gc_train['target'] = y_train.values
                    synth.fit(X_gc_train)
                    X_synth = synth.sample(len(X_train_enc))
                    y_tr = X_synth['target']
                    X_tr = X_synth.drop(columns=['target'])
                else:
                    print(" gaussian copula not available; skipping")
                    continue
            except Exception as e:
                print(" gaussian copula error:", e)
                continue
        else:
            print(" unknown method", method)
            continue

        # ensure DataFrame and target series names
        if not isinstance(X_tr, pd.DataFrame):
            X_tr = pd.DataFrame(X_tr, columns=X_train_enc.columns)
        if not isinstance(y_tr, pd.Series):
            y_tr = pd.Series(y_tr, name='target')

        train_data = pd.concat([X_tr.reset_index(drop=True), y_tr.reset_index(drop=True)], axis=1)
        train_data.rename(columns={train_data.columns[-1]: 'target'}, inplace=True)

        # fit an AutoGluon predictor for this run (use small models set or same set as you used)
        predictor = TabularPredictor(label='target', eval_metric='roc_auc').fit(
            train_data,
            hyperparameters={'XGB' : {}, 'RF' : {}, 'GBM' : {}, 'NN_TORCH' : {}, 'CAT' : {}},
            verbosity=0
        )

        # evaluate on X_test_enc
        X_test_df = X_test_enc.copy()
        y_true = y_test.values
        # predictions
        y_pred = predictor.predict(X_test_df).values
        # probabilities: predictor.predict_proba returns DataFrame with columns for classes (0 and 1)
        try:
            proba_df = predictor.predict_proba(X_test_df)
            if isinstance(proba_df, pd.DataFrame):
                # try to access column '1' or last column
                if 1 in proba_df.columns:
                    y_proba = proba_df[1].values
                else:
                    y_proba = proba_df.iloc[:, -1].values
            else:
                y_proba = np.array(proba_df)[:, 1]
        except Exception:
            # fallback: get predicted labels as probs (0/1)
            y_proba = y_pred.astype(float)

        # metrics
        try:
            roc = roc_auc_score(y_true, y_proba)
        except Exception:
            roc = np.nan

        # for fairness we want per-protected-attribute results; we compute aggregates (max differences) across all categorical_features
        # compute Equalized Odds aggregated: we compute per-protected attr and then aggregate by taking max TPR and FPR diffs across all provided attrs
        eo_TPR_diffs = []
        eo_FPR_diffs = []
        di_max_devs = []
        pp_diffs = []
        for prot in categorical_features:
            protected_attr_test = X_sensitive_test[prot].reset_index(drop=True)
            eo = equalized_odds_difference(y_true, y_pred, protected_attr_test)
            di = disparate_impact(y_pred, protected_attr_test)
            pp = predictive_parity(y_true, y_pred, protected_attr_test)
            if not np.isnan(eo['TPR_difference']):
                eo_TPR_diffs.append(eo['TPR_difference'])
            if not np.isnan(eo['FPR_difference']):
                eo_FPR_diffs.append(eo['FPR_difference'])
            if not np.isnan(di['max_abs_deviation_from_1']):
                di_max_devs.append(di['max_abs_deviation_from_1'])
            if not np.isnan(pp['PP_difference']):
                pp_diffs.append(pp['PP_difference'])

        # aggregate per-run (we use max across sensitive attributes to be conservative)
        run_record = {
            'roc_auc': roc,
            'eo_TPR_diff': np.nanmax(eo_TPR_diffs) if eo_TPR_diffs else np.nan,
            'eo_FPR_diff': np.nanmax(eo_FPR_diffs) if eo_FPR_diffs else np.nan,
            'di_max_abs_dev': np.nanmax(di_max_devs) if di_max_devs else np.nan,
            'pp_diff': np.nanmax(pp_diffs) if pp_diffs else np.nan,
            'method': method
        }

        results_by_method[method].append(run_record)

elapsed = time.time() - start_time
print(f"\nFinished all runs in {elapsed:.1f}s")

# -----------------------------------------
# --- Summarize: mean & std per method ---
# -----------------------------------------
summary_rows = []
for method, runs in results_by_method.items():
    if not runs:
        continue
    df_runs = pd.DataFrame(runs)
    summary = {
        'method': method,
        'n_runs': len(df_runs),
        'roc_auc_mean': df_runs['roc_auc'].mean(),
        'roc_auc_std': df_runs['roc_auc'].std(),
        'eo_TPR_diff_mean': df_runs['eo_TPR_diff'].mean(),
        'eo_TPR_diff_std': df_runs['eo_TPR_diff'].std(),
        'eo_FPR_diff_mean': df_runs['eo_FPR_diff'].mean(),
        'eo_FPR_diff_std': df_runs['eo_FPR_diff'].std(),
        'di_max_abs_dev_mean': df_runs['di_max_abs_dev'].mean(),
        'di_max_abs_dev_std': df_runs['di_max_abs_dev'].std(),
        'pp_diff_mean': df_runs['pp_diff'].mean(),
        'pp_diff_std': df_runs['pp_diff'].std()
    }
    summary_rows.append(summary)

summary_df = pd.DataFrame(summary_rows).sort_values('roc_auc_mean', ascending=False).reset_index(drop=True)
pd.options.display.float_format = '{:0.4f}'.format
print("\nSummary (mean ± std across runs):")
display(summary_df)

# Save summary to CSV for reporting
summary_df.to_csv('synthetic_methods_summary_stats.csv', index=False)
print("\nSaved summary to synthetic_methods_summary_stats.csv")
```

```python
# Dependency preservation scoring (one block to paste into your notebook)
import pandas as pd
import numpy as np
from scipy.stats import spearmanr
from numpy.linalg import norm

# --- Helper: safe checker for variable existence in globals() ---
env = globals()  # uses notebook globals so run this in same notebook

# Map likely variable names (based on your notebook)
candidate_map = {
    'Baseline': 'X_encoded',          # original one-hot matrix
    'SMOTE': 'X_resampled',           # SMOTE result (array or DF)
    'CTGAN': 'X_synthetic',           # CTGAN produced X_synthetic in your code
    'TVAE': 'synthetic_data',         # TVAE used variable name synthetic_data in your code
    'Gaussian Copula': 'synthetic_data_copula'  # sometimes synthetic_data reused; set fallback below
}

# Build actual dictionary of DataFrames for synthetic datasets
synthetic_datasets = {}

# Baseline (original features) - ensure DataFrame
if 'X_encoded' in env:
    X_orig = env['X_encoded'].copy()
    if not isinstance(X_orig, pd.DataFrame):
        X_orig = pd.DataFrame(X_orig, columns=env.get('X_encoded_columns', None))
else:
    raise RuntimeError("X_encoded (original one-hot features) not found. Make sure it exists and run this cell after preprocessing.")

# Helper to coerce to DataFrame and align columns (one-hot)
def make_df(obj, ref_cols=None):
    if obj is None:
        return None
    if isinstance(obj, pd.DataFrame):
        df = obj.copy()
    elif isinstance(obj, (np.ndarray, list)):
        # If numpy array, try to attach ref_cols if provided
        if ref_cols is not None and len(ref_cols) == obj.shape[1]:
            df = pd.DataFrame(obj, columns=ref_cols)
        else:
            df = pd.DataFrame(obj)
    else:
        # unknown type, try to convert
        try:
            df = pd.DataFrame(obj)
        except Exception:
            return None
    # ensure same column order as reference where possible
    if ref_cols is not None:
        # keep columns that match, reindex missing as NaN
        common = [c for c in ref_cols if c in df.columns]
        if len(common) == 0:
            # possibly the synthetic is raw categorical; try to one-hot it
            try:
                df = pd.get_dummies(df, drop_first=True)
            except Exception:
                pass
        # reindex to ref cols, filling missing with zeros
        df = df.reindex(columns=ref_cols, fill_value=0)
    return df

# Gather datasets by checking variables present in the environment
# For Gaussian Copula / TVAE your notebook sometimes used 'synthetic_data' variable multiple times.
# We'll check a few possible names used in the notebook.
possible_names = {
    'SMOTE': ['X_resampled', 'X_resampled_smote'],
    'CTGAN': ['X_synthetic', 'X_synth', 'ctgan_synthetic', 'X_synthetic_ctgan'],
    'TVAE': ['synthetic_data', 'synthetic_data_tvae', 'tvae_synth'],
    'Gaussian Copula': ['synthetic_data', 'synthetic_data_copula', 'copula_synthetic', 'gc_synthetic']
}

# First put baseline
synthetic_datasets['Baseline'] = make_df(X_orig, ref_cols=list(X_orig.columns))

# For each method try to find a variable
for method, names in possible_names.items():
    found = False
    for nm in names:
        if nm in env:
            df_candidate = make_df(env[nm], ref_cols=list(X_orig.columns))
            if df_candidate is not None:
                synthetic_datasets[method] = df_candidate
                found = True
                break
    # if not found but method == 'Gaussian Copula' and there is 'synthetic_data' check if it's from copula by shape
    if (not found) and ('synthetic_data' in env) and (method not in synthetic_datasets):
        # use it as fallback (user overwrote variable in a later block)
        synthetic_datasets[method] = make_df(env['synthetic_data'], ref_cols=list(X_orig.columns))

# If SMOTE present only as arrays (X_resampled, y_resampled used earlier), we already attempt to capture it above.
# Final sanity: ensure all stored are DataFrames and have same columns as baseline
for k, v in list(synthetic_datasets.items()):
    if v is None:
        synthetic_datasets.pop(k, None)
    else:
        # coerce columns numeric (Spearman needs numeric)
        synthetic_datasets[k] = v.astype(float, errors='ignore').fillna(0.0)

if len(synthetic_datasets) == 0:
    raise RuntimeError("No synthetic datasets found in the environment. Check variable names or run generation cells first.")

# --- Define dependency preservation score using Spearman correlation matrix similarity ---
def spearman_corr_matrix(df):
    # compute Spearman correlation matrix (pandas will handle NaNs)
    if df.shape[1] < 2:
        # trivial case
        return np.eye(df.shape[1])
    corr, _ = spearmanr(df, axis=0, nan_policy='omit')
    # spearmanr returns a full matrix sized (n_features, n_features) when input is 2D
    # if df has 1-D or small, ensure shape
    if isinstance(corr, float) or corr is None:
        return np.eye(df.shape[1])
    # If corr is larger (like when df had rows<cols), ensure correct shape:
    corr = np.asarray(corr)
    # Clip to [-1,1]
    corr = np.clip(corr, -1.0, 1.0)
    return corr

def preservation_score(R_orig, R_synth):
    # compute Frobenius norm of difference normalized by original norm
    num = norm(R_orig - R_synth, ord='fro')
    denom = norm(R_orig, ord='fro') if norm(R_orig, ord='fro') > 0 else 1.0
    score = 1.0 - (num / denom)
    # clamp to sensible range
    if np.isnan(score):
        return np.nan
    return float(np.clip(score, -1.0, 1.0))

# Compute scores
results = []
R_orig = spearman_corr_matrix(synthetic_datasets['Baseline'])
for method, df_synth in synthetic_datasets.items():
    if method == 'Baseline':
        score = 1.0
    else:
        # keep only numeric columns that exist in both (reindex done earlier)
        R_synth = spearman_corr_matrix(df_synth)
        # ensure same shape
        if R_synth.shape != R_orig.shape:
            # try to reindex columns (shouldn't happen if reindexing above succeeded)
            min_dim = min(R_orig.shape[0], R_synth.shape[0])
            R_orig_trim = R_orig[:min_dim, :min_dim]
            R_synth_trim = R_synth[:min_dim, :min_dim]
            score = preservation_score(R_orig_trim, R_synth_trim)
        else:
            score = preservation_score(R_orig, R_synth)
    results.append({'method': method, 'dependency_preservation': score,
                    'n_features': synthetic_datasets[method].shape[1],
                    'n_rows': synthetic_datasets[method].shape[0]})

results_df = pd.DataFrame(results).sort_values(by='dependency_preservation', ascending=False).reset_index(drop=True)

# Print and display neatly for copy-paste into Google Docs
print("Dependency preservation (Spearman-based) — one score per method")
print(results_df.to_string(index=False))

# Also output DataFrame object in case you want to reuse it programmatically
results_df
```
